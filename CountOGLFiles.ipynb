```python
import os
import re
modsurls = {}
archive_urls = {}
hathi_urls = {}
slub_urls = {}
gbooks_urls = {}
unique_bookurls = {}
basebook_urls = {}
pagelinks = 0

def additem(curitem,curdict):
    if(curitem in curdict):
        curdict[curitem] = curdict[curitem] + 1
    else:
        curdict[curitem] = 1

def getpdfs(fname):

    global modsurls
    global pagelinks
    curpdfs = []
    f=open(fname)
    for l in f:
        m = re.search('<mods:url[^>]*>([^<]+)',l)
        if(m):
            cururl = m[1]
            if(re.search('(archive.org|books.google|hdl.handle.net|slub-dresden)',cururl)):
                if(re.search('#.+',cururl)):
                    pagelinks = pagelinks + 1
                baseurl = re.sub('#.+','',cururl)
                additem( baseurl,basebook_urls)
            if(re.search('archive.org',cururl)):
                additem(cururl,archive_urls)
                curpdfs.append(cururl)
            if(re.search('books.google',cururl)):
                additem(cururl,gbooks_urls)
                curpdfs.append(cururl)
            if(re.search('hdl.handle.net',cururl)):
                additem(cururl,hathi_urls)
                curpdfs.append(cururl)
            if(re.search('slub-dresden',cururl)):
                additem(cururl,slub_urls)
                curpdfs.append(cururl)
            if(cururl in modsurls):
                modsurls[cururl] = modsurls[cururl] + 1
            else:
                modsurls[cururl] = 1
    f.close()
    return(curpdfs)
```
