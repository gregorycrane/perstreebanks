```python
tglist = {}
editionlist = {}

def dorepo(currepo):
    for top, dirs, files in os.walk(currepo):
        for file in files:
            if(not re.search('data/',os.path.join(top, file))):
                continue
            m = re.search('([a-z]+[0-9][0-9][0-9][0-9])\.([a-z]+[0-9][0-9][0-9])\.([^\-]+\-(grc|eng|lat|deu|fra)([0-9]+))\.xml',file)
            if(m):
                curtgid = m[1]
                curworkid = m[2]
                curedition = m[3]
                curlang = m[4]
                curversion = m[5]
                curkey = curtgid + '.' + curworkid
                if(curtgid in tglist):
                    if(not curworkid in tglist[curtgid]):
                        tglist[curtgid].append(curworkid)
                else:
                    tglist[curtgid] = [curworkid]
                if(curkey in editionlist):
                    editionlist[curkey].append(os.path.join(top, file))
                    #if(re.search('tlg0527',curkey)):
                        #print('huh?',os.path.join(top, file),curkey,editionlist[curkey])
                else:
                    editionlist[curkey] = [os.path.join(top, file)]
            #print(curtgid,curworkid,curedition,curlang,curversion,file)

def checkdupes(curl):
    sawdupes = 0
    tmps = ' '.join(curl)
    if(re.search('tlg7000',tmps)):
        return(sawdupes)
    #print(tmps)
    if(re.search('eng1',tmps) and re.search('eng2',tmps)):
        for foo in curl:
            if(re.search('-eng1',foo)):
                #curl.remove(foo)
                #print('removed',foo,curl)
                sawdupes = sawdupes + 1
                tmps = ' '.join(curl)

    if(re.search('opp-grc',tmps) and re.search('1st1K-grc[12]',tmps)):
        for foo in curl:
            if(re.search('opp-grc',foo)):
                curl.remove(foo)
                sawdupes = sawdupes + 1
                #print('removed',foo,curl)
                tmps = ' '.join(curl)

                continue
    if(re.search('grc1',tmps) and re.search('grc2',tmps)):
        for foo in curl:
            if(re.search('-grc1',foo)):
                curl.remove(foo)
                sawdupes = sawdupes + 1
                #print('removed2',foo,curl)
        #print('dupe',curl)
    return(sawdupes)
  
# filter out dupes where grc2 and eng2 are just reformatted versions of 
# grc1 and eng1

dorepo(repodir)
dorepo(persdir)
for foo in editionlist:
    startl = len(editionlist[foo])
    rval = checkdupes(editionlist[foo])
#    if(re.search('0527',foo)):
#        print('sept',foo,editionlist[foo])
    #if(rval):
    #    print(startl,len(editionlist[foo]),editionlist[foo])
    
    

```

```python

def getauthwork(fname):
    retauth = ''
    retwork = ''
    f = open(fname)
    for l in f:
        if(re.search('<title>Machine readable text</title>',l)):
            continue
        if(retauth and retwork):
            break
        m = re.search('<author[^>]*>([^<]+)',l)
        if(not retauth and m):
            retauth = m[1]
            #print('author',retauth)
        m = re.search('<title\\b[^>]*>([^<]+)',l)
        if(not retwork and m):
            retwork = m[1]
            if(re.search('New Testament',retwork)):
                retauth,retwork = retwork.split(' - ')
                break
            if(re.search('IOANNIS ZONARAE EPITOME HISTORIARUM',retwork)):
                retauth = 'Joannes Zonaras'
            if(re.search('Scholia Euripidem',retwork)):
                retwork = 'Scholia in Euripidem'
               # print(retauth,retwork)
                break
            #print('work',retwork,l)
    f.close()
    retwork = re.sub('\s+$','',retwork)
    retwork = re.sub('^\s+','',retwork)
    retwork = re.sub('\s+\((English|Greek|GREEK|ENGLISH)\)\.*','',retwork)
    retwork = re.sub('\s+Machine readable( text|)\.*','',retwork)
    retwork = re.sub('\s+from$','',retwork)
    retwork = re.sub('World English Bible[\s\-]+','',retwork)
    return(retauth,retwork)

def check4xlation(curwork):
    if(re.search('-eng',' '.join(editionlist[curwork]))):
        return('has_xlation')
    else:
        return('no_xlation')
# now calculate word lengths for each remaining work file


editionsizes = {}
authorworknames = {}
worklist = []
fromrepo = ''
for foo in editionlist:
    curtgid,curworkid = foo.split('.')
    for boo in editionlist[foo]:
        #fname = repodir + '/data/' + curtgid + '/'  + curworkid + '/' + boo
        fname = boo
        print(fname)
        editionsizes[boo] = countwords(fname)
        #print(boo,editionsizes[boo])
        if(not foo in authorworknames or 
           (re.search('[α-ωΑ-Ω]',authorworknames[foo]) and re.search('-eng',fname))):
            curauth,curwork = getauthwork(fname)
            authorworknames[foo] = curwork
            if(boo in editionsizes):
                cursize = editionsizes[boo]
            else:
                cursize = 0
        if( curtgid in genretable):
            prose_poetry = genretable[curtgid]
        else:
            prose_poetry = 'prose'
        has_xlation = check4xlation(foo)
        if(re.search('perseus',boo)):
            fromrepo = 'perseus'
        else:
            fromrepo = 'first1k'
    worklist.append([curtgid,tg_namelist[curtgid],curworkid,authorworknames[foo],cursize,prose_poetry,re.sub(' .+','',workclasses[foo]),has_xlation,fromrepo])
    #print('aw',authorworknames)     

authorworknames
```
