```python
#authorlemmas['tlg0012']['μῆνις']
import re
authorlemmas = {}
def lemauthfreqs(curlem):
    prosetot = 0
    poetrytot = 0
    wordprosetot = 0
    wordpoetrytot = 0
    curlem = re.sub('-','',curlem)
    curtot = 0
    reslist = {}
    global authorlemmas
    for foo in authorlemmas:
        if(foo in authinfo):
            fullinfo = authinfo[foo]
            curinfo = authinfo[foo].split(',')[0]
        else:
            curinfo = 'noinfo'
            fullinfo = ''
                         
        if(re.search('poetry',fullinfo)):
            poetrytot = poetrytot + authorwordcnt[foo]
        else:
            prosetot = prosetot + authorwordcnt[foo]
        
        if(curlem in authorlemmas[foo] ):
            currat = authorlemmas[foo][curlem] * (10000/authorwordcnt[foo])
            curtot = curtot + authorlemmas[foo][curlem]
            if(re.search('poetry',fullinfo)):
                wordpoetrytot = wordpoetrytot + authorlemmas[foo][curlem]
            else:
                wordprosetot = wordprosetot + authorlemmas[foo][curlem]
            #print(currat,foo,authorlemmas[foo][curlem],authorwordcnt[foo],curinfo)
            reslist[foo] = currat
        else:
            reslist[foo] = 0
            #print(0,foo,curinfo,'no hits')
    for foo in sorted(reslist,key=reslist.get,reverse=True):
        if(foo in authinfo):
            curinfo = authinfo[foo].split(',')[0]
        else:
            curinfo = 'noinfo'
        if(curlem in authorlemmas[foo] ):
            curcnt = authorlemmas[foo][curlem]
        else:
            curcnt = 0
        #print(curlem,"{:.2f}".format(reslist[foo]),foo,curcnt,authorwordcnt[foo],curinfo)
    print(curlem,curtot,'prose',wordprosetot,"{:.2f}".format((10000*wordprosetot)/prosetot),'poetry',wordpoetrytot,"{:.2f}".format((10000*wordpoetrytot)/poetrytot))
```
