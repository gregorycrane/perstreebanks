```python
import os
import re

dirname = '/Users/gcrane/github/german_trans-dev/'
dirname = '/Users/gcrane/github/nicenefathers'
dirname = '/Users/gcrane/github/italian_trans-dev/'
dirname = '/Users/gcrane/github/french_trans-dev/'
dirname = '/Users/gcrane/github/german_trans-dev/'
dirname = '/Users/gcrane/github/english_trans-dev/volumes'

nfiles = 0

def countwords(fname):
    curcnt = 0
    f = open(fname)
    for l in f:
        
        l = re.sub('<[^>]+>',' ',l)
        curcnt = curcnt + len(l.split())
    f.close()
    return(curcnt)
    

totwords = 0
for top, dirs, files in os.walk(dirname):
    #print("Printing directories...")
    #for dir in dirs:
    #    print(os.path.join(top, dir))
    #print("Printing files....")
    #print(len(files))
    for file in files:
        if(not re.search('\.xml$',file)):
            continue
        nfiles = nfiles + 1
        curwds = countwords(os.path.join(top, file))
        totwords = curwds + totwords
        print(nfiles,curwds,totwords,os.path.join(top, file))
```
